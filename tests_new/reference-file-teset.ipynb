```python
from hwpx import HwpxDocument
```

```python
with HwpxDocument.open("./input/reference-testfile.hwpx") as doc:
    sections = doc.sections
```

```python
sections, len(sections)
```

```python
[print(para.text) for para in sections[0].paragraphs]
print()
```

```python
[
    [  # [c for r in nlist for c in r]
        [print(f"para{i}: {[cell.cell.text for cell in row]}") for row in table.get_cell_map()]
        for table in para.tables
    ]
    for i, para in enumerate(sections[0].paragraphs) if para.tables
]
print()
```

```python
from tabulate import tabulate

tables = \
[
    [
        tabulate([[cell.cell.text for cell in row] for row in table.get_cell_map()], tablefmt='github')
        for table in para.tables
    ] 
    for i, para in enumerate(sections[0].paragraphs) if para.tables
]

[print(table, '\n\n') for table_l in tables for table in table_l]
print()
```

```python
tables_raw = \
[
    [
        [[cell.cell for cell in row] for row in table.get_cell_map()]
        for table in para.tables
    ] 
    for para in sections[0].paragraphs if para.tables
]
```
